```python
from abc import ABC, abstractmethod
from datetime import datetime
from typing import Dict, List, Optional


class CurrencyMismatchError(Exception):
    """Raised when an operation is attempted across mismatched currencies without conversion."""
    pass


class InsufficientFundsError(Exception):
    """Raised when an account does not have sufficient funds for a withdrawal/fee."""
    pass


# ==========================================
# 1. ABSTRACTION & ENCAPSULATION: Base Model
# ==========================================
class Transaction:
    """Represents an immutable record of a financial transaction."""

    def __init__(self, transaction_type: str, amount: float, currency: str, balance_after: float):
        self.__timestamp = datetime.now()
        self.__transaction_type = transaction_type
        self.__amount = amount
        self.__currency = currency
        self.__balance_after = balance_after

    @property
    def timestamp(self) -> datetime:
        return self.__timestamp

    @property
    def transaction_type(self) -> str:
        return self.__transaction_type

    @property
    def amount(self) -> float:
        return self.__amount

    @property
    def currency(self) -> str:
        return self.__currency

    @property
    def balance_after(self) -> float:
        return self.__balance_after

    def __repr__(self) -> str:
        return (f"[{self.__timestamp.strftime('%Y-%m-%d %H:%M:%S')}] "
                f"{self.__transaction_type:<18} | Amount: {self.__amount:10.2f} {self.__currency} "
                f"| Balance: {self.__balance_after:10.2f} {self.__currency}")


class Account(ABC):
    """Abstract Base Class defining the blueprint for all bank accounts."""

    def __init__(self, account_id: str, owner_name: str, currency: str = "USD", initial_balance: float = 0.0):
        # Strict validation
        if not account_id or not isinstance(account_id, str):
            raise ValueError("Account ID must be a non-empty string.")
        if not owner_name or not isinstance(owner_name, str):
            raise ValueError("Owner name must be a non-empty string.")
        if initial_balance < 0:
            raise ValueError("Initial balance cannot be negative.")

        # Encapsulation: private attributes
        self.__account_id: str = account_id
        self.__owner_name: str = owner_name
        self.__currency: str = currency.upper()
        self.__balance: float = float(initial_balance)
        self.__transaction_history: List[Transaction] = []

        if initial_balance > 0:
            self._record_transaction("INITIAL_DEPOSIT", initial_balance)

    # --- Encapsulation: Getters and Controlled Properties ---
    @property
    def account_id(self) -> str:
        return self.__account_id

    @property
    def owner_name(self) -> str:
        return self.__owner_name

    @property
    def currency(self) -> str:
        return self.__currency

    @property
    def balance(self) -> float:
        return self.__balance

    @property
    def transaction_history(self) -> List[Transaction]:
        # Return a copy to preserve internal state integrity
        return list(self.__transaction_history)

    def _set_balance(self, new_balance: float) -> None:
        """Internal helper for derived classes to adjust balance with validation."""
        self.__balance = round(new_balance, 2)

    def _record_transaction(self, tx_type: str, amount: float) -> None:
        """Internal ledger record creation."""
        tx = Transaction(tx_type, amount, self.__currency, self.__balance)
        self.__transaction_history.append(tx)

    # --- Core Operations ---
    def deposit(self, amount: float) -> None:
        if amount <= 0:
            raise ValueError("Deposit amount must be strictly positive.")
        self._set_balance(self.balance + amount)
        self._record_transaction("DEPOSIT", amount)

    def withdraw(self, amount: float) -> None:
        if amount <= 0:
            raise ValueError("Withdrawal amount must be strictly positive.")
        if self.balance < amount:
            raise InsufficientFundsError(
                f"Insufficient funds: Balance is {self.balance} {self.currency}, requested {amount} {self.currency}."
            )
        self._set_balance(self.balance - amount)
        self._record_transaction("WITHDRAWAL", amount)

    # --- Abstraction: Methods that must be implemented by concrete subclasses ---
    @abstractmethod
    def accrue_interest(self) -> float:
        """Calculates, applies, and returns periodic interest."""
        pass

    @abstractmethod
    def deduct_fees(self) -> float:
        """Calculates, applies, and returns account maintenance/overdraft fees."""
        pass

    def __str__(self) -> str:
        return f"{self.__class__.__name__}(ID={self.account_id}, Owner='{self.owner_name}', Balance={self.balance:.2f} {self.currency})"


# ==========================================
# 2. INHERITANCE: Concrete Account Subclasses
# ==========================================
class SavingsAccount(Account):
    """Standard savings account offering moderate interest and flat maintenance fee."""

    def __init__(self, account_id: str, owner_name: str, currency: str = "USD",
                 initial_balance: float = 0.0, annual_interest_rate: float = 0.03):
        super().__init__(account_id, owner_name, currency, initial_balance)
        if annual_interest_rate < 0:
            raise ValueError("Interest rate cannot be negative.")
        self.__interest_rate: float = annual_interest_rate
        self.__maintenance_fee: float = 5.0

    def accrue_interest(self) -> float:
        interest = round(self.balance * (self.__interest_rate / 12), 2)
        if interest > 0:
            self._set_balance(self.balance + interest)
            self._record_transaction("INTEREST_ACCRUED", interest)
        return interest

    def deduct_fees(self) -> float:
        fee = self.__maintenance_fee if self.balance < 500.0 else 0.0
        if fee > 0:
            if self.balance < fee:
                raise InsufficientFundsError(f"Cannot deduct fee: balance {self.balance} is below fee {fee}.")
            self._set_balance(self.balance - fee)
            self._record_transaction("MAINTENANCE_FEE", fee)
        return fee


class CheckingAccount(Account):
    """Transactional account with overdraft buffer and overdraft transaction fees."""

    def __init__(self, account_id: str, owner_name: str, currency: str = "USD",
                 initial_balance: float = 0.0, overdraft_limit: float = 500.0):
        super().__init__(account_id, owner_name, currency, initial_balance)
        if overdraft_limit < 0:
            raise ValueError("Overdraft limit cannot be negative.")
        self.__overdraft_limit: float = overdraft_limit
        self.__overdraft_fee: float = 25.0

    @property
    def overdraft_limit(self) -> float:
        return self.__overdraft_limit

    def withdraw(self, amount: float) -> None:
        if amount <= 0:
            raise ValueError("Withdrawal amount must be positive.")
        available = self.balance + self.__overdraft_limit
        if amount > available:
            raise InsufficientFundsError(
                f"Exceeds overdraft limit. Available: {available:.2f} {self.currency}, Requested: {amount:.2f} {self.currency}."
            )

        self._set_balance(self.balance - amount)
        self._record_transaction("WITHDRAWAL", amount)

    def accrue_interest(self) -> float:
        # Standard checking accounts typically do not accrue interest
        return 0.0

    def deduct_fees(self) -> float:
        fee = 0.0
        if self.balance < 0:
            fee = self.__overdraft_fee
            self._set_balance(self.balance - fee)
            self._record_transaction("OVERDRAFT_FEE", fee)
        return fee


class HighYieldAccount(Account):
    """High-yield investment/savings account requiring minimum threshold for premium APY."""

    def __init__(self, account_id: str, owner_name: str, currency: str = "USD",
                 initial_balance: float = 0.0, apy: float = 0.07, minimum_threshold: float = 2500.0):
        super().__init__(account_id, owner_name, currency, initial_balance)
        self.__apy: float = apy
        self.__minimum_threshold: float = minimum_threshold
        self.__penalty_fee: float = 35.0

    def accrue_interest(self) -> float:
        if self.balance >= self.__minimum_threshold:
            interest = round(self.balance * (self.__apy / 12), 2)
            self._set_balance(self.balance + interest)
            self._record_transaction("HIGH_YIELD_INTEREST", interest)
            return interest
        return 0.0

    def deduct_fees(self) -> float:
        fee = 0.0
        if self.balance < self.__minimum_threshold:
            fee = self.__penalty_fee
            if self.balance < fee:
                raise InsufficientFundsError("Insufficient funds for minimum balance penalty.")
            self._set_balance(self.balance - fee)
            self._record_transaction("LOW_BALANCE_PENALTY", fee)
        return fee


# ==========================================
# 3. POLYMORPHISM: Financial Ledger Engine
# ==========================================
class LedgerEngine:
    """Manages multi-currency accounts and polymorphically dispatches transactions."""

    def __init__(self):
        self.__accounts: Dict[str, Account] = {}

    def register_account(self, account: Account) -> None:
        if account.account_id in self.__accounts:
            raise ValueError(f"Account with ID {account.account_id} already exists.")
        self.__accounts[account.account_id] = account

    def get_account(self, account_id: str) -> Account:
        if account_id not in self.__accounts:
            raise KeyError(f"Account {account_id} not found in ledger.")
        return self.__accounts[account_id]

    def transfer(self, source_id: str, destination_id: str, amount: float) -> None:
        """Dispatches an atomic transfer between two registered accounts."""
        source = self.get_account(source_id)
        target = self.get_account(destination_id)

        if source.currency != target.currency:
            raise CurrencyMismatchError(
                f"Direct transfer failed: {source.currency} to {target.currency}. FX conversion is required."
            )

        # Polymorphic withdrawal and deposit
        source.withdraw(amount)
        target.deposit(amount)
        print(f"[TRANSFER OK] {amount:.2f} {source.currency} from {source_id} -> {destination_id}")

    def run_end_of_month_cycle(self) -> None:
        """Polymorphically runs interest accrual and fee deductions across all accounts."""
        print("\n" + "=" * 60)
        print("EXECUTING END-OF-MONTH SETTLEMENT BATCH (POLYMORPHIC DISPATCH)")
        print("=" * 60)

        for account in self.__accounts.values():
            # Polymorphic execution: each subclass handles fees & interest per its own rules
            fees = account.deduct_fees()
            interest = account.accrue_interest()
            print(f"[{account.account_id} - {type(account).__name__}] Fees Charged: {fees:6.2f} | Interest Accrued: {interest:6.2f} | Final Balance: {account.balance:9.2f} {account.currency}")


# ==========================================
# Verification & Execution
# ==========================================
if __name__ == "__main__":
    engine = LedgerEngine()

    # Create instances of different account types
    sav = SavingsAccount(account_id="SAV-101", owner_name="Alice Smith", currency="USD", initial_balance=400.0)
    chk = CheckingAccount(account_id="CHK-202", owner_name="Bob Jones", currency="USD", initial_balance=150.0, overdraft_limit=300.0)
    hya = HighYieldAccount(account_id="HYA-303", owner_name="Charlie Lee", currency="USD", initial_balance=10000.0)

    # Register in the ledger engine
    engine.register_account(sav)
    engine.register_account(chk)
    engine.register_account(hya)

    # Perform polymorphic transactions
    print("\n--- Initial State ---")
    for acc_id in ["SAV-101", "CHK-202", "HYA-303"]:
        print(engine.get_account(acc_id))

    # Test checking overdraft behavior
    print("\n--- Overdraft Operation ---")
    chk.withdraw(250.0)  # Account will go negative (-100.00 USD), leveraging overdraft buffer
    print(f"Bob's checking balance after overdraft: {chk.balance:.2f} {chk.currency}")

    # Transfer funds polymorphically
    print("\n--- Inter-Account Transfer ---")
    engine.transfer(source_id="HYA-303", destination_id="SAV-101", amount=1200.0)

    # Run month-end batch processing
    engine.run_end_of_month_cycle()

    # Print transaction history from the internal ledger
    print("\n" + "=" * 60)
    print("AUDIT TRAIL: SAVINGS ACCOUNT (SAV-101)")
    print("=" * 60)
    for record in sav.transaction_history:
        print(record)
```

```python
from abc import ABC, abstractmethod
from dataclasses import dataclass
from typing import List


# =====================================================================
# 1. Single Responsibility Principle (SRP): Pure Data Representation
# =====================================================================
@dataclass(frozen=True)
class NotificationMessage:
    """Immutable data transfer object representing the payload."""
    recipient: str
    subject: str
    body: str


# =====================================================================
# 2. Interface Segregation (ISP) & Liskov Substitution (LSP): Abstraction
# =====================================================================
class NotificationChannel(ABC):
    """Abstract interface defining the contract for any notification medium."""

    @abstractmethod
    def send(self, message: NotificationMessage) -> bool:
        """Sends the message to the intended recipient via the specific provider channel."""
        pass

    @property
    @abstractmethod
    def channel_name(self) -> str:
        """Returns the human-readable identifier of the notification provider."""
        pass


# =====================================================================
# 3. Open/Closed Principle (OCP): Concrete Channel Implementations
# =====================================================================
class EmailNotificationChannel(NotificationChannel):
    """Email provider dispatch adapter."""

    @property
    def channel_name(self) -> str:
        return "Email (SMTP/SendGrid)"

    def send(self, message: NotificationMessage) -> bool:
        if "@" not in message.recipient:
            print(f"[{self.channel_name}] Failed: Invalid email format -> {message.recipient}")
            return False
        print(f"[{self.channel_name}] -> To: {message.recipient} | Subject: '{message.subject}' | Body: '{message.body}'")
        return True


class SMSNotificationChannel(NotificationChannel):
    """SMS provider dispatch adapter (e.g., Twilio)."""

    @property
    def channel_name(self) -> str:
        return "SMS (Twilio)"

    def send(self, message: NotificationMessage) -> bool:
        # Validate international or local phone format
        if not (message.recipient.startswith("+") and message.recipient[1:].isdigit()):
            print(f"[{self.channel_name}] Failed: Invalid phone number -> {message.recipient}")
            return False
        print(f"[{self.channel_name}] -> To: {message.recipient} | SMS Content: '{message.body}'")
        return True


class SlackNotificationChannel(NotificationChannel):
    """Slack webhook / API notification adapter."""

    @property
    def channel_name(self) -> str:
        return "Slack Webhook"

    def send(self, message: NotificationMessage) -> bool:
        # Expecting a webhook URL or channel name
        print(f"[{self.channel_name}] -> Channel: {message.recipient} | Alert: [{message.subject}] {message.body}")
        return True


class WhatsAppNotificationChannel(NotificationChannel):
    """WhatsApp Business API notification adapter."""

    @property
    def channel_name(self) -> str:
        return "WhatsApp Business API"

    def send(self, message: NotificationMessage) -> bool:
        if not (message.recipient.startswith("+") and message.recipient[1:].isdigit()):
            print(f"[{self.channel_name}] Failed: Invalid WhatsApp number -> {message.recipient}")
            return False
        print(f"[{self.channel_name}] -> WhatsApp To: {message.recipient} | Message: '{message.body}'")
        return True


# =====================================================================
# 4. Dependency Inversion Principle (DIP): High-Level Notification Engine
# =====================================================================
class NotificationDispatcher:
    """High-level dispatch engine decoupled from concrete channel implementations."""

    def __init__(self, channels: List[NotificationChannel] = None):
        # Dependency Injection via constructor
        self._channels: List[NotificationChannel] = channels if channels is not None else []

    def register_channel(self, channel: NotificationChannel) -> None:
        """Allows runtime dynamic registration without modifying core class (OCP)."""
        if not isinstance(channel, NotificationChannel):
            raise TypeError("Supplied channel must implement NotificationChannel interface.")
        self._channels.append(channel)

    def dispatch_broadcast(self, message: NotificationMessage) -> dict:
        """Broadcasts a single notification payload across all registered channels."""
        print(f"\n--- Broadcasting Notification: '{message.subject}' ---")
        results = {}
        for channel in self._channels:
            success = channel.send(message)
            results[channel.channel_name] = success
        return results

    def dispatch_targeted(self, channel_name: str, message: NotificationMessage) -> bool:
        """Dispatches an alert to a specific channel identified by name."""
        for channel in self._channels:
            if channel.channel_name.lower() == channel_name.lower():
                print(f"\n--- Dispatching Targeted Alert via {channel.channel_name} ---")
                return channel.send(message)
        print(f"Targeted dispatch failed: Channel '{channel_name}' not registered.")
        return False


# =====================================================================
# Verification & Demonstration
# =====================================================================
if __name__ == "__main__":
    # 1. Dependency Injection: Initialize concrete channels
    email_channel = EmailNotificationChannel()
    sms_channel = SMSNotificationChannel()
    slack_channel = SlackNotificationChannel()
    whatsapp_channel = WhatsAppNotificationChannel()

    # 2. Inject channels into the dispatch service
    dispatcher = NotificationDispatcher(channels=[
        email_channel,
        sms_channel,
        slack_channel,
        whatsapp_channel
    ])

    # 3. Create test alerts
    system_alert = NotificationMessage(
        recipient="devops-alerts@datheon.in",
        subject="High CPU Utilization Detected",
        body="Server us-east-1 load exceeded 95% for 5 consecutive minutes."
    )

    customer_sms = NotificationMessage(
        recipient="+919876543210",
        subject="One-Time Password",
        body="Your OTP for login verification is 482910."
    )

    slack_alert = NotificationMessage(
        recipient="#incident-room",
        subject="Deployment Success",
        body="Service 'ledger-core' successfully deployed to production."
    )

    whatsapp_alert = NotificationMessage(
        recipient="+919876543210",
        subject="Order Shipped",
        body="Your hardware token order has been dispatched via BlueDart."
    )

    # 4. Demonstrate targeted dispatching
    dispatcher.dispatch_targeted("Email (SMTP/SendGrid)", system_alert)
    dispatcher.dispatch_targeted("SMS (Twilio)", customer_sms)
    dispatcher.dispatch_targeted("Slack Webhook", slack_alert)
    dispatcher.dispatch_targeted("WhatsApp Business API", whatsapp_alert)

    # 5. Demonstrate Extensibility (OCP): Plug in a brand new channel without changing engine
    class TelegramNotificationChannel(NotificationChannel):
        """Newly created Telegram Bot channel."""
        @property
        def channel_name(self) -> str:
            return "Telegram Bot"

        def send(self, message: NotificationMessage) -> bool:
            print(f"[{self.channel_name}] -> Chat ID: {message.recipient} | Text: '{message.body}'")
            return True

    print("\n--- Testing Open/Closed Principle (Adding Telegram Channel Dynamically) ---")
    telegram_channel = TelegramNotificationChannel()
    dispatcher.register_channel(telegram_channel)

    telegram_msg = NotificationMessage(
        recipient="@datheon_broadcast_bot",
        subject="Security Update",
        body="New patch installed successfully across edge gateways."
    )
    dispatcher.dispatch_targeted("Telegram Bot", telegram_msg)
```

```python
from abc import ABC, abstractmethod
from enum import Enum
import threading
from typing import List, Dict, Any
from datetime import datetime


# =====================================================================
# 1. SINGLETON PATTERN: Thread-Safe Database & Configuration Manager
# =====================================================================
class DatabaseConfigManager:
    """Thread-safe Singleton implementation managing central database & app configuration."""
    _instance = None
    _lock = threading.Lock()

    def __new__(cls):
        with cls._lock:
            if cls._instance is None:
                cls._instance = super(DatabaseConfigManager, cls).__new__(cls)
                cls._instance._init_config()
        return cls._instance

    def _init_config(self):
        self.connection_string = "postgresql://cluster_admin:secure_pass@db.internal:5432/ecommerce_prod"
        self.max_retries = 3
        self.currency = "USD"
        self.active_connections = 0
        print("[Singleton] DatabaseConfigManager instance created and configured.")

    def log_query(self, query: str):
        print(f"[DB LOG - {datetime.now().strftime('%H:%M:%S')}] {query}")


# =====================================================================
# 2. FACTORY METHOD PATTERN: Polymorphic Payment Handlers
# =====================================================================
class PaymentMethod(Enum):
    CREDIT_CARD = "credit_card"
    UPI = "upi"
    PAYPAL = "paypal"
    CRYPTO = "crypto"


class PaymentHandler(ABC):
    """Abstract Product defining interface for all payment providers."""
    @abstractmethod
    def process_payment(self, amount: float, order_id: str) -> bool:
        pass


class CreditCardHandler(PaymentHandler):
    def process_payment(self, amount: float, order_id: str) -> bool:
        print(f"[Payment Gateway] Processing Credit Card payment of ${amount:.2f} for Order #{order_id} via Stripe.")
        return True


class UPIHandler(PaymentHandler):
    def process_payment(self, amount: float, order_id: str) -> bool:
        print(f"[Payment Gateway] Processing UPI instant settlement of ${amount:.2f} for Order #{order_id} via VPA.")
        return True


class PayPalHandler(PaymentHandler):
    def process_payment(self, amount: float, order_id: str) -> bool:
        print(f"[Payment Gateway] Processing PayPal Express Checkout of ${amount:.2f} for Order #{order_id}.")
        return True


class CryptoHandler(PaymentHandler):
    def process_payment(self, amount: float, order_id: str) -> bool:
        print(f"[Payment Gateway] Verifying crypto blockchain transfer of ${amount:.2f} for Order #{order_id}.")
        return True


class PaymentHandlerFactory:
    """Factory creating instances of payment handlers dynamically."""
    _handlers = {
        PaymentMethod.CREDIT_CARD: CreditCardHandler,
        PaymentMethod.UPI: UPIHandler,
        PaymentMethod.PAYPAL: PayPalHandler,
        PaymentMethod.CRYPTO: CryptoHandler,
    }

    @classmethod
    def get_handler(cls, method: PaymentMethod) -> PaymentHandler:
        handler_cls = cls._handlers.get(method)
        if not handler_cls:
            raise ValueError(f"Unsupported payment method: {method}")
        return handler_cls()


# =====================================================================
# 3. OBSERVER PATTERN: Event Listeners for Order Lifecycle
# =====================================================================
class OrderStatus(Enum):
    PLACED = "PLACED"
    PAID = "PAID"
    PACKED = "PACKED"
    DISPATCHED = "DISPATCHED"


class OrderObserver(ABC):
    """Subscriber interface for order status broadcast."""
    @abstractmethod
    def update(self, order_id: str, status: OrderStatus, details: Dict[str, Any]) -> None:
        pass


class LogisticsService(OrderObserver):
    def update(self, order_id: str, status: OrderStatus, details: Dict[str, Any]) -> None:
        if status == OrderStatus.PAID:
            print(f" [Logistics] Order #{order_id} verified. Generating shipping label for {details.get('address')}.")
        elif status == OrderStatus.DISPATCHED:
            print(f" [Logistics] Order #{order_id} picked up by carrier. Tracking: TRK-{order_id}-X.")


class InvoiceService(OrderObserver):
    def update(self, order_id: str, status: OrderStatus, details: Dict[str, Any]) -> None:
        if status == OrderStatus.PAID:
            print(f" [Invoicing] PDF Tax invoice generated for Order #{order_id}. Total: ${details.get('amount'):.2f}.")


class EmailNotificationService(OrderObserver):
    def update(self, order_id: str, status: OrderStatus, details: Dict[str, Any]) -> None:
        print(f" [Notification] Email alert dispatched to '{details.get('email')}': Order #{order_id} is now [{status.value}].")


# =====================================================================
# ORDER PIPELINE (Subject Orchestrator)
# =====================================================================
class OrderPipeline:
    """Subject orchestrating order execution and notifying registered observers."""

    def __init__(self, order_id: str, customer_email: str, shipping_address: str, amount: float):
        self.order_id = order_id
        self.customer_email = customer_email
        self.shipping_address = shipping_address
        self.amount = amount
        self.status = OrderStatus.PLACED
        self._observers: List[OrderObserver] = []
        self.db = DatabaseConfigManager()  # Singleton call

    def attach_observer(self, observer: OrderObserver) -> None:
        if observer not in self._observers:
            self._observers.append(observer)

    def detach_observer(self, observer: OrderObserver) -> None:
        self._observers.remove(observer)

    def _notify(self) -> None:
        payload = {
            "email": self.customer_email,
            "address": self.shipping_address,
            "amount": self.amount
        }
        for observer in self._observers:
            observer.update(self.order_id, self.status, payload)

    def set_status(self, new_status: OrderStatus) -> None:
        self.status = new_status
        self.db.log_query(f"UPDATE orders SET status = '{new_status.value}' WHERE id = '{self.order_id}';")
        self._notify()

    def process_order(self, method: PaymentMethod) -> bool:
        print(f"\n==================== PROCESSING ORDER #{self.order_id} ====================")
        self._notify()  # Initial PLACED status broadcast

        # Factory Pattern execution
        handler = PaymentHandlerFactory.get_handler(method)
        payment_success = handler.process_payment(self.amount, self.order_id)

        if payment_success:
            self.set_status(OrderStatus.PAID)
            self.set_status(OrderStatus.PACKED)
            self.set_status(OrderStatus.DISPATCHED)
            return True
        return False


# =====================================================================
# Execution & Verification
# =====================================================================
if __name__ == "__main__":
    # Test Singleton behavior
    db1 = DatabaseConfigManager()
    db2 = DatabaseConfigManager()
    assert db1 is db2, "Singleton test failed!"
    print(f"Verified: db1 and db2 point to identical instance in memory: {db1 is db2}\n")

    # Initialize order pipeline
    order = OrderPipeline(
        order_id="ORD-9482",
        customer_email="rahul.intern@datheon.in",
        shipping_address="Sector 62, Noida, UP, India",
        amount=1499.00
    )

    # Register Observers
    order.attach_observer(LogisticsService())
    order.attach_observer(InvoiceService())
    order.attach_observer(EmailNotificationService())

    # Execute payment using Factory
    order.process_order(PaymentMethod.UPI)
```

```python
from abc import ABC
from enum import Enum
from datetime import datetime, timedelta
import math
from typing import Dict, List, Optional


# =====================================================================
# 1. Enums: Types, Sizes, and Dynamic Rates
# =====================================================================
class VehicleType(Enum):
    MOTORCYCLE = "MOTORCYCLE"
    COMPACT_CAR = "COMPACT_CAR"
    LARGE_SUV = "LARGE_SUV"
    ELECTRIC_VEHICLE = "ELECTRIC_VEHICLE"


class SpotType(Enum):
    MOTORCYCLE = "MOTORCYCLE"
    COMPACT = "COMPACT"
    LARGE = "LARGE"
    ELECTRIC = "ELECTRIC"


# =====================================================================
# 2. Vehicle Hierarchy
# =====================================================================
class Vehicle(ABC):
    def __init__(self, license_plate: str, vehicle_type: VehicleType):
        self.license_plate = license_plate
        self.vehicle_type = vehicle_type


class Motorcycle(Vehicle):
    def __init__(self, license_plate: str):
        super().__init__(license_plate, VehicleType.MOTORCYCLE)


class CompactCar(Vehicle):
    def __init__(self, license_plate: str):
        super().__init__(license_plate, VehicleType.COMPACT_CAR)


class LargeSUV(Vehicle):
    def __init__(self, license_plate: str):
        super().__init__(license_plate, VehicleType.LARGE_SUV)


class ElectricVehicle(Vehicle):
    def __init__(self, license_plate: str):
        super().__init__(license_plate, VehicleType.ELECTRIC_VEHICLE)


# =====================================================================
# 3. Parking Spot Model
# =====================================================================
class ParkingSpot:
    def __init__(self, spot_id: str, spot_type: SpotType):
        self.spot_id = spot_id
        self.spot_type = spot_type
        self.is_occupied: bool = False
        self.current_vehicle: Optional[Vehicle] = None

    def can_fit_vehicle(self, vehicle: Vehicle) -> bool:
        """Determines if a vehicle fits the designated spot size/type."""
        if self.is_occupied:
            return False
        if vehicle.vehicle_type == VehicleType.MOTORCYCLE:
            return True  # Motorcycle fits in any spot
        if vehicle.vehicle_type == VehicleType.COMPACT_CAR:
            return self.spot_type in [SpotType.COMPACT, SpotType.LARGE]
        if vehicle.vehicle_type == VehicleType.LARGE_SUV:
            return self.spot_type == SpotType.LARGE
        if vehicle.vehicle_type == VehicleType.ELECTRIC_VEHICLE:
            return self.spot_type == SpotType.ELECTRIC
        return False

    def assign_vehicle(self, vehicle: Vehicle) -> None:
        self.current_vehicle = vehicle
        self.is_occupied = True

    def vacate(self) -> None:
        self.current_vehicle = None
        self.is_occupied = False


# =====================================================================
# 4. Parking Floor
# =====================================================================
class ParkingFloor:
    def __init__(self, floor_number: int):
        self.floor_number = floor_number
        self.spots: List[ParkingSpot] = []

    def add_spot(self, spot: ParkingSpot) -> None:
        self.spots.append(spot)

    def find_available_spot(self, vehicle: Vehicle) -> Optional[ParkingSpot]:
        """Finds the best matching slot available on this floor."""
        for spot in self.spots:
            if spot.can_fit_vehicle(vehicle):
                return spot
        return None


# =====================================================================
# 5. Ticket & Dynamic Rate Calculator
# =====================================================================
class ParkingTicket:
    def __init__(self, ticket_id: str, vehicle: Vehicle, spot: ParkingSpot, floor_number: int):
        self.ticket_id = ticket_id
        self.vehicle = vehicle
        self.spot = spot
        self.floor_number = floor_number
        self.entry_time = datetime.now()


class BillingEngine:
    """Dynamic hourly pricing calculation based on vehicle type and electricity charging."""
    BASE_HOURLY_RATES = {
        VehicleType.MOTORCYCLE: 15.0,
        VehicleType.COMPACT_CAR: 30.0,
        VehicleType.LARGE_SUV: 50.0,
        VehicleType.ELECTRIC_VEHICLE: 40.0,  # Base rate + charging infrastructure
    }

    @classmethod
    def calculate_fee(cls, ticket: ParkingTicket, exit_time: datetime) -> float:
        duration = exit_time - ticket.entry_time
        # Minimum billable unit is 1 hour
        hours = max(1, math.ceil(duration.total_seconds() / 3600))
        rate = cls.BASE_HOURLY_RATES.get(ticket.vehicle.vehicle_type, 30.0)
        return float(hours * rate)


# =====================================================================
# 6. Central Smart Parking Lot Manager
# =====================================================================
class SmartParkingLot:
    def __init__(self, name: str):
        self.name = name
        self.floors: List[ParkingFloor] = []
        self.active_tickets: Dict[str, ParkingTicket] = {}
        self._ticket_counter = 1000

    def add_floor(self, floor: ParkingFloor) -> None:
        self.floors.append(floor)

    def park_vehicle(self, vehicle: Vehicle) -> Optional[ParkingTicket]:
        # Allocation algorithm: search floor by floor for the first eligible slot
        for floor in self.floors:
            spot = floor.find_available_spot(vehicle)
            if spot:
                spot.assign_vehicle(vehicle)
                self._ticket_counter += 1
                ticket_id = f"TICK-{self._ticket_counter}"
                ticket = ParkingTicket(ticket_id, vehicle, spot, floor.floor_number)
                self.active_tickets[ticket_id] = ticket
                print(f"[PARKED] {vehicle.vehicle_type.value} ({vehicle.license_plate}) "
                      f"-> Floor {floor.floor_number}, Spot {spot.spot_id} | Ticket: {ticket_id}")
                return ticket

        print(f"[REJECTED] Parking full for {vehicle.vehicle_type.value} ({vehicle.license_plate})!")
        return None

    def unpark_vehicle(self, ticket_id: str, simulated_hours: int = 1) -> float:
        ticket = self.active_tickets.get(ticket_id)
        if not ticket:
            raise KeyError(f"Ticket #{ticket_id} is invalid or expired.")

        # Simulate exit time for testing
        exit_time = ticket.entry_time + timedelta(hours=simulated_hours)
        fee = BillingEngine.calculate_fee(ticket, exit_time)

        # Free the slot
        ticket.spot.vacate()
        del self.active_tickets[ticket_id]

        print(f"[UNPARKED] {ticket.vehicle.license_plate} freed Spot {ticket.spot.spot_id} (Floor {ticket.floor_number}) "
              f"| Duration: {simulated_hours} hr(s) | Fee: ₹{fee:.2f}")
        return fee


# =====================================================================
# Verification & Test Cases
# =====================================================================
if __name__ == "__main__":
    # 1. Setup multi-floor parking lot
    lot = SmartParkingLot("CyberHub Smart Garage")

    # Floor 1: Ground floor with Electric & Motorcycle slots
    floor1 = ParkingFloor(1)
    floor1.add_spot(ParkingSpot("F1-M1", SpotType.MOTORCYCLE))
    floor1.add_spot(ParkingSpot("F1-E1", SpotType.ELECTRIC))
    lot.add_floor(floor1)

    # Floor 2: Compact and Large SUV spots
    floor2 = ParkingFloor(2)
    floor2.add_spot(ParkingSpot("F2-C1", SpotType.COMPACT))
    floor2.add_spot(ParkingSpot("F2-L1", SpotType.LARGE))
    lot.add_floor(floor2)

    print("\n--- Testing Automated Slot Allocation ---")
    bike = Motorcycle("UP-14-BK-1102")
    ev = ElectricVehicle("DL-01-EV-9999")
    suv = LargeSUV("HR-26-SUV-0001")
    compact = CompactCar("DL-08-CC-4321")

    t1 = lot.park_vehicle(bike)
    t2 = lot.park_vehicle(ev)
    t3 = lot.park_vehicle(suv)
    t4 = lot.park_vehicle(compact)

    print("\n--- Testing Allocation Under Capacity Constraint ---")
    extra_suv = LargeSUV("DL-03-SUV-8888")
    lot.park_vehicle(extra_suv)  # Should reject because Floor 2 Large spot is occupied

    print("\n--- Testing Unparking & Dynamic Rate Calculation ---")
    if t1:
        lot.unpark_vehicle(t1.ticket_id, simulated_hours=3)   # 3 hours for bike (₹15/hr = ₹45)
    if t2:
        lot.unpark_vehicle(t2.ticket_id, simulated_hours=2)   # 2 hours for EV (₹40/hr = ₹80)
    if t3:
        lot.unpark_vehicle(t3.ticket_id, simulated_hours=4)   # 4 hours for SUV (₹50/hr = ₹200)

    print("\n--- Re-attempting Previously Rejected Vehicle ---")
    lot.park_vehicle(extra_suv)  # Ab Floor 2 ka large slot free ho gaya, allocation success hoga
```
